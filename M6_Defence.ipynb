## Likely defence questions and concise answers

**What is the base paper?** Jay Barach, ResumeShield, arXiv:2609.20188v1.

**Did you reproduce its exact implementation?** No. M2 is a local reconstruction; the corpus wording, the success definition and the filters differ, and 2 of six pipeline rows do not match the paper.

**What is new in your work?** Not an algorithm. Visibility comparison and injection classifiers both exist. New is the controlled evaluation of their combination, with ablations, on a real LLM and real PDF files, with the prediction recorded first.

**What was the hypothesis and was it supported?** At least 20 points lower ASR with at least 95% clean agreement. Actual: 75.0 points (21/24 to 3/24) and 100.0%. Supported, on this sample.

**How do we know the prediction came first?** The protocol file is timestamped before the first model response and stores hashes of the inputs. It is a local record, not an external preregistration.

**Are there 24 independent trials?** No. They come from 4 resumes, and the variants of one resume are correlated. The interval of 58.3 to 83.3 points is descriptive.

**Why Llama 3.2 3B and not GPT-4o-mini or Gemini?** No hosted model was usable without payment; one refused the free tier and one stopped at 140 of 210 calls. The hypothesis and thresholds were identical in all three protocol files.

**Which attacks still succeed?** 3/24: the visible paraphrase. It is really on the page, so the visibility check has nothing to flag, and the pretrained classifier scores it near zero.

**Why did channel separation do worse than in the paper?** In the paper the simulator never follows fenced text. A real small model still followed the direct override in 9/12 cases, and the JSON wrapping changed a clean decision.

**Does the fine-tuned classifier fix the paraphrase?** It catches every payload block and flags no genuine block here, but it is exploratory: its training sentences were written after seeing the misses, and its effect on attack success has not been measured.

**Could the defence remove genuine resume text?** In this run it removed 30 blocks, 30 of them payloads and 0 from clean documents. The sample is small.

**How good is the scanner?** On unseen resumes it found 350 of 350 hidden-text attacks. Under the current rules it passed 139 of 140 clean files and called 39 of 60 non-hidden instructions an attack. Visible instructions in unusual wording are its weak case.

**What about the earlier pilot?** A sentence-level TF-IDF detector. Recall passed and the false-positive rate did not, so its hypothesis was not supported. It is reported as a negative result and its numbers are not mixed with the ASR results.

**Is this safe to deploy?** The evidence does not show that. Six resumes, one small model, two payload wordings. A flag should prompt a person to look, never reject a candidate automatically.

**What would you do next?** A new frozen run with the fine-tuned classifier, new payload wording by another author, a second resume source, a second annotator and a larger model.
