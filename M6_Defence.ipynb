## Likely defence questions and concise answers

**What is the base paper?** Jay Barach, ResumeShield, arXiv:2609.20188v1.

**Did you reproduce its exact implementation?** No. M2 is a local reconstruction; corpus wording, success definition and filters differ.

**What did you train?** A standard logistic-regression detector over TF-IDF text features. No LLM was trained.

**What is new?** The small frozen family-split evaluation, trained-detector comparison, ablations and evidence-based negative result; not a new algorithm.

**Why did your hypothesis fail?** Recall passed, but three of sixteen benign test sources were flagged, violating the FPR limit.

**Why is keyword recall zero?** The test attacks stress unfamiliar phrasing; the narrow list misses them. That limits generalization of the comparison.

**Are there 96 independent test examples?** No: 32 source sentences, each with three related encodings.

**How did you avoid leakage?** Families and source sentences remain in one split; TF-IDF fits only on training; thresholds use validation only. Single-author semantic overlap remains a limitation.

**Does high recall mean fewer successful attacks?** Not necessarily. We measured detection, not outcomes on a real LLM.

**Why not change the threshold now?** That would tune to the test set. A revised model needs a new held-out test.

**Can you submit a failed hypothesis?** Yes: the rubric asks for controlled evidence and honest analysis, not a guaranteed improvement.
